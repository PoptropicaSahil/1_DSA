```python
from typing import List
```

```python
# QUESTION : Given an array having both positive and negative integers.Find length of the largest subarray with sum 0.
# Simlar to https://leetcode.com/problems/contiguous-array/description/
# (SOLVED)

class Solution:
    def longestSubarray(self, nums: List[int]) -> int:

        sum_map = {}  # key : cumsum,  value: index
        max_len = 0   
        cum_sum = 0  

        for i in range(len(nums)):
            cum_sum += nums[i]  # Update the cumulative sum
            
            # If cum_sum is 0, that means the subarray from index 0 to i has a sum of 0
            if cum_sum == 0:
                max_len = i + 1
            
            # If found the same cumsum again, then middle part has zero sum
            if cum_sum in sum_map:
                max_len = max(max_len, i - sum_map[cum_sum])
            else:
                sum_map[cum_sum] = i

        return max_len
```
