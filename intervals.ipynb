```python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:

        for i in range(len(nums)-1):

            print(f'now i={i}')
            if i < len(nums):
                cur_char = nums[i]
            else: break

            if i+1 < len(nums) and nums[i+1] == cur_char:
                while i+2 < len(nums) and nums[i+2] == cur_char:
                    nums.pop(i+2)
            else:
                print(f'going in else for i={i}')
                continue
            
            print(f'now i= {i}, nums = {nums}')

        return 1



(
    Solution().removeDuplicates([0, 0, 1, 1, 1, 1, 2, 3, 3]),
    Solution().removeDuplicates([1, 1, 1, 2, 2, 3]),
)
        
```

```python
class Solution:
    def duplicateZeros(self, arr: List[int]) -> None:
        """
        Do not return anything, modify arr in-place instead.
        """
        
        i = 0
        
        while True:

            if i >= len(arr): 
                break

            if arr[i] == 0:
                arr.insert(i+1, 0)
                i += 2

            else:
                i += 1

            print(f"now arr = {arr}")


(
    Solution().duplicateZeros([1, 0, 2, 3, 0, 4, 5, 0]),
    Solution().duplicateZeros([4, 0, 0, 0, 2, 3, 0, 4, 5, 0]),
)
```

```python
class Solution:
    def sampleStats(self, count: List[int]) -> List[float]:
        self.count = count
        self.mean = self._mean()

        return [self.mean, self.mean]

    def _mean(self) -> float:
        # summ = 0
        # cnt = 0
        # for idx, n in enumerate(self.count):
        #     if n != 0:
        #         summ += idx * n
        #         cnt += n
        
        # if cnt == 0:
        #     return 0
        
        # return summ / cnt

        summ =  sum([e*i for i,e in enumerate(self.count)])
        cnt = sum(self.count)

        if cnt == 0:
            return 0

        return summ / cnt

    
    def _mode(self) -> int:
        max_count = max(self.count)
        return self.count.index(max_count)

    def _minimum(self) -> int:
        for i in range(len(self.count)):
            if self.count[i] != 0:
                return i
        return 0
    
    def _maximum(self) -> int:
        for i in range(len(self.count)-1, -1, -1):
            if self.count[i] != 0:
                return i
        return 0
    
    def _median(self) -> float:
        total_count = sum(self.count)
        mid = total_count // 2
        count_so_far = 0
        
        # If total_count is odd, we need to find the middle element
        if total_count % 2 == 1:
            for i in range(len(self.count)):
                count_so_far += self.count[i]
                if count_so_far > mid:
                    return i
        
        # If total_count is even, we need to find the two middle elements
        else:
            first = None
            second = None
            for i in range(len(self.count)):
                count_so_far += self.count[i]
                if count_so_far > mid:
                    second = i
                    if first is None:
                        first = i
                    return (first + second) / 2
                if count_so_far == mid:
                    first = i

        return 0
```

```python
class Solution:
    def leastBricks(self, wall: List[List[int]]) -> int:

        cum_wall = [[sum(w[:i]) for i in range(1, len(w))] for w in wall]
        res = len(wall) # number of rows
        width = sum(wall[0])

        for i in range(1, width):
            cuts_by_this = 0
            for row in cum_wall:
                if i not in row:
                    cuts_by_this += 1
            print(f'for i={i}, cuts_by_this = {cuts_by_this}')
            res = min(res, cuts_by_this)
        
        
        return res
    
Solution().leastBricks([[1,2,2,1],[3,1,2],[1,3,2],[2,4],[3,1,2],[1,3,1,1]])
```

```python
class Solution:
    def findPeakElement(self, nums: List[int]) -> int:
        n = len(nums)
        nums = [-float('inf')] + nums + [-float('inf')]
        print(f'nums={nums}')

        for i in range(1, n):
            print(f'checking i={i}, nums={nums}, nums i = {nums[i]}')
            if nums[i-1] < nums[i] and nums[i+1] < nums[i]:

                return i-1 # idx wrt original
        

# Solution().findPeakElement([3, 1, 2]), 
Solution().findPeakElement([3])
```

```python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        if k> len(nums):
            k = k%len(nums)
        k = len(nums) % k
        while k:
            last = nums.pop()
            print(f'last = {last}, nums = {nums}')
            nums = [last] + nums
            k -= 1
            print(f'k = {k}, nums ipdated = {nums}')

Solution().rotate([1,2,3,4,5,6,7], 3)
```

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        if len(nums) == 1:
            if nums[0] >= target: return 1
            else: return 0

        # prefix_sum = [sum(nums[:i]) for i in range(len(nums)+1) ]
        # print(f'prefix_sum = {prefix_sum}')
        
        l, r = 0, 0
        n = len(nums)

        res = n+1

        while l <= n and r <= n:
            # cur_sum = prefix_sum[r] - prefix_sum[l-1]
            cur_sum = sum(nums[l: r+1])

            print(f'l={l}, r={r}, cur_sum = {cur_sum}')
            # pass

            if cur_sum < target:
                r += 1
                print(f'\t cur_sum < target, r = {r}')
            elif cur_sum >= target:
                res = min(res, r - l + 1)
                l += 1
                print(f'\t cur_sum >= target, l = {l}, res = {res}')


        if res == n+1:
            return 0
        
        return res

# Solution().minSubArrayLen(7, [2,3,1,2,4,3])
# Solution().minSubArrayLen(4, [1,4,4])
# Solution().minSubArrayLen(11, [1, 1, 1, 1, 1, 1, 1, 1])
Solution().minSubArrayLen(6, [10, 2, 3])
```
